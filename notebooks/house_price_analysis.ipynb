# House Price Prediction - Exploratory Data Analysis

This notebook explores the existing `Housing.csv` dataset. It summarizes the data and reports its quality and categorical values. No data cleaning or model training is performed.

## 1. Load and inspect the dataset

The CSV is stored in the project's `dataset` folder. This relative path is written from the `notebooks` folder.

```python
from pathlib import Path

import pandas as pd

data_path = Path("../dataset/Housing.csv")
houses = pd.read_csv(data_path)

print(f"Dataset file: {data_path.name}")
print(f"Dataset shape: {houses.shape[0]} rows and {houses.shape[1]} columns")
print("Column names:", houses.columns.tolist())
```

These tables show the first and last five houses in the file.

## 3. Dataset structure and data types

The shape gives the number of records and columns. `info()` lists each column's data type and non-missing value count.

```python
print("Dataset shape:", houses.shape)
print("\nColumn names:")
print(houses.columns.tolist())
print("\nData types:")
print(houses.dtypes)
print("\nDataset information:")
houses.info()
```

## 4. Statistical summary

`describe(include="all")` summarizes numeric columns and also reports counts and common values for categorical columns.

```python
houses.describe(include="all")
```

## 5. Missing values and duplicate records

The per-column totals help identify missing entries. The duplicate count reports rows that are identical across all columns; no rows are removed here.

```python
missing_by_column = houses.isna().sum()
total_missing_values = int(missing_by_column.sum())
duplicate_records = int(houses.duplicated().sum())

print("Missing values in each column:")
print(missing_by_column)
print("\nTotal missing values:", total_missing_values)
print("Duplicate records:", duplicate_records)
```

## 6. Unique values in categorical columns

The categorical columns contain text categories such as yes/no indicators and furnishing status. This cell lists the values observed in each one.

```python
categorical_columns = houses.select_dtypes(include=["object", "category", "string"]).columns

for column in categorical_columns:
    print(f"{column}: {houses[column].unique().tolist()}")
```

## 7. Identify the target variable

Inspection of the actual column names shows that `price` is the house price and is the target variable for a future prediction task.

```python
target_column = "price"
if target_column not in houses.columns:
    raise KeyError(f"Expected target column {target_column!r} was not found in the dataset.")

print("Target variable:", target_column)
```

## 8. EDA findings

- The file is `Housing.csv` and contains **545 records** and **13 columns**.
- The target variable is **`price`**. The other columns describe house area, rooms, amenities, parking, preferred area, and furnishing status.
- There are **6 numeric columns** and **7 categorical columns**. The categorical features include yes/no values and three furnishing-status categories.
- The dataset has **no missing values** and **no duplicate records**.
- The `price` values range from **1,750,000** to **13,300,000** in the source dataset's units.

These are observations only. The dataset has not been cleaned or transformed, and no model has been trained.
